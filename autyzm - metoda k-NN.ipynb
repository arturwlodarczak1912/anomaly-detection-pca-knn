```python
import numpy as np
import pandas as pd
```

```python
df = df.drop(columns=['ethnicity', 'relation'])
```

```python
df = df.drop(columns=['contry_of_res',	'used_app_before',	'age_desc',	'Class/ASD', 'result'], errors='ignore')
```

```python
df = df[df['age'] < 200]
```

```python
df.isnull().sum().sum()
```

```python
df = pd.get_dummies(df, drop_first=True)
df
```

## Standaryzacja danych (zadanie)

```python
df_dummy = df
```

```python
X, y = df_dummy.drop(columns=['austim_yes']), df_dummy['austim_yes']
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state=402)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
df_scaled = scaler.fit(X_train)  #  ważne skalujemy tylko na zbiorze traninigowym
```

```python
X_train  = df_scaled.transform(X_train)
X_test = df_scaled.transform(X_test)
```

```python
from sklearn import neighbors
```

```python
knn = neighbors.KNeighborsClassifier(n_neighbors=3)
```

```python
knn.fit(X_train, y_train)
```

```python
knn.score(X_test, y_test), knn.score(X_train, y_train)
```

```python
from sklearn.metrics import confusion_matrix
```

```python
confusion_matrix(y_test, knn.predict(X_test))
```

```python
from sklearn.metrics import classification_report
```

```python
print(classification_report(y_test, knn.predict(X_test)))
```

```python
print(classification_report(y_train, knn.predict(X_train)))
```

# Model: random forest

```python
from sklearn import ensemble
rfc = ensemble.RandomForestClassifier()
```

```python
rfc.fit(X_train, y_train)
```

```python
print(classification_report(y_test, rfc.predict(X_test)))
```

# Model logistic regression

```python
from sklearn.linear_model import LogisticRegression
```

```python
clf = LogisticRegression(random_state=0, C=2)
```

```python
clf = LogisticRegression(random_state=0).fit(X_train, y_train)
```

```python
print(classification_report(y_test, clf.predict(X_test)))
```

# Model tree

```python
from sklearn import tree
clf_tree = tree.DecisionTreeClassifier(max_depth=7)  #  max_depth to jest przykład hiper parametru
clf_tree.fit(X_train, y_train)
print(classification_report(y_test, clf_tree.predict(X_test)))
```

```python
print(classification_report(y_train, clf_tree.predict(X_train)))
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=402, train_size=0.7)
```

```python
scaler = StandardScaler()
scaler.fit(X_train)
```

```python
X_train = scaler.transform(X_train)
X_test = scaler.transform(X_test)
```

```python
X_test.shape
```

```python
y_test.mean(), y_train.mean()
```

```python
from sklearn import tree
clf_tree = tree.DecisionTreeClassifier(max_depth=14)  #  max_depth to jest przykład hiper parametru
clf_tree.fit(X_train, y_train)
clf_tree.score(X_test, y_test), clf_tree.score(X_train, y_train)
print(classification_report(y_test, clf_tree.predict(X_test)))
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
clf = LogisticRegression(random_state=462, C=1 )
clf.fit(X_train, y_train)
clf.score(X_test, y_test), clf.score(X_train, y_train)
print(classification_report(y_test, clf.predict(X_test)))
```
